### PEAS - Data structures and fringes that define the Agent environment goes here

```python
import numpy as np
import random

# ============================================================
#  PEAS - Data Structures and Fringes that define the Agent environment
# ============================================================

ROWS = 6
COLS = 7
PLAYER = 1     # Represents DISK OF PLAYER 1
BOT = 2   # Represents DISK of computer
EMPTY = 0
SET_COUNT = 4  # 4 set of same disk.


# Initialize board 7X6 (COLXROW) with all 0
def create_board():
    board = np.zeros((ROWS, COLS))
    return board

# Drop colored disk into the board.
def drop_disk(board, row, col, disk):
    board[row][col] = disk

# Display the board state
def display_board_state(board):
    print(np.flip(board, 0))

# Check if the location is empty or not.
def is_location_valid(board, col):
    return board[ROWS-1][col] == 0

# Get next empty row
def get_next_empty_row(board, col):
    for r in range(ROWS):
        if board[r][col] == 0:
            return r

# Get all columns where a move is possible
def get_valid_locations(board):
    valid_locations = [col for col in range(COLS) if is_location_valid(board, col)]
    return valid_locations

def winning_move(board, disk):
    # Horizontal check
    for c in range(COLS - 3):
        for r in range(ROWS):
            if all([board[r][c+i] == disk for i in range(SET_COUNT)]):
                return True
    # Vertical check
    for c in range(COLS):
        for r in range(ROWS - 3):
            if all([board[r+i][c] == disk for i in range(SET_COUNT)]):
                return True
    # Check for digonals
    for c in range(COLS - 3):
        for r in range(ROWS - 3):
            if all([board[r+i][c+i] == disk for i in range(SET_COUNT)]):
                return True
    for c in range(COLS - 3):
        for r in range(3, ROWS):
            if all([board[r-i][c+i] == disk for i in range(SET_COUNT)]):
                return True
    return False

def is_leaf_node(board):
    return winning_move(board, PLAYER) or winning_move(board, BOT) or len(get_valid_locations(board)) == 0
```

### Implementation of the Min-Max algorithm

```python
def minimax(board, depth, maximizingPlayer):
    
    valid_locations = get_valid_locations(board)
    is_terminal = is_leaf_node(board)

    if depth == 0 or is_terminal:
        if is_terminal:
            if winning_move(board, BOT):
                return (None, 100000000)
            elif winning_move(board, PLAYER):
                return (None, -100000000)
            else:
                return (None, 0)
        else:
            return (None, score_position(board, BOT))

    if maximizingPlayer:
        value = -np.inf
        column = random.choice(valid_locations)
        for col in valid_locations:
            row = get_next_empty_row(board, col)
            temp_board = board.copy()
            drop_disk(temp_board, row, col, BOT)
            new_score = minimax(temp_board, depth-1, False)[1]
            if new_score > value:
                value = new_score
                column = col
        return column, value
    else:
        value = np.inf
        column = random.choice(valid_locations)
        for col in valid_locations:
            row = get_next_empty_row(board, col)
            temp_board = board.copy()
            drop_disk(temp_board, row, col, PLAYER)
            new_score = minimax(temp_board, depth-1, True)[1]
            if new_score < value:
                value = new_score
                column = col
        return column, value

```

### Implementation of the alpha-beta pruning  

```python
#Code block
def minimax_alpha_beta(board, depth, alpha, beta, maximizingPlayer):
    valid_locations = get_valid_locations(board)
    is_terminal = is_leaf_node(board)

    if depth == 0 or is_terminal:
        if is_terminal:
            if winning_move(board, BOT):
                return (None, 100000000)
            elif winning_move(board, PLAYER):
                return (None, -100000000)
            else:
                return (None, 0)
        else:
            return (None, score_position(board, BOT))

    if maximizingPlayer:
        value = -np.inf
        column = random.choice(valid_locations)
        for col in valid_locations:
            row = get_next_empty_row(board, col)
            temp_board = board.copy()
            drop_disk(temp_board, row, col, BOT)
            new_score = minimax_alpha_beta(temp_board, depth-1, alpha, beta, False)[1]
            if new_score > value:
                value = new_score
                column = col
            alpha = max(alpha, value)
            if alpha >= beta:
                break  # Alpha prune
        return column, value
    else:
        value = np.inf
        column = random.choice(valid_locations)
        for col in valid_locations:
            row = get_next_empty_row(board, col)
            temp_board = board.copy()
            drop_disk(temp_board, row, col, PLAYER)
            new_score = minimax_alpha_beta(temp_board, depth-1, alpha, beta, True)[1]
            if new_score < value:
                value = new_score
                column = col
            beta = min(beta, value)
            if alpha >= beta:
                break  # Beta prune
        return column, value
```

### Choice and implementation of the Static Evaluation Function.

```python
#Code block - Start the game
def score_position(board, disk):

    score = 0

    # Score center column higher (central control is advantageous)
    center_array = [int(i) for i in list(board[:, COLS // 2])]
    center_count = center_array.count(disk)
    score += center_count * 3

    # Score Horizontal
    for r in range(ROWS):
        row_array = [int(i) for i in list(board[r,:])]
        for c in range(COLS - 3):
            window = row_array[c:c + SET_COUNT]
            score += evaluate_window(window, disk)

    # Score Vertical
    for c in range(COLS):
        col_array = [int(i) for i in list(board[:, c])]
        for r in range(ROWS - 3):
            window = col_array[r:r + SET_COUNT]
            score += evaluate_window(window, disk)

    # Score positively sloped diagonals
    for r in range(ROWS - 3):
        for c in range(COLS - 3):
            window = [board[r + i][c + i] for i in range(SET_COUNT)]
            score += evaluate_window(window, disk)

    # Score negatively sloped diagonals
    for r in range(ROWS - 3):
        for c in range(COLS - 3):
            window = [board[r + 3 - i][c + i] for i in range(SET_COUNT)]
            score += evaluate_window(window, disk)

    return score


def evaluate_window(window, disk):

    score = 0
    opp_disk = PLAYER if disk == BOT else BOT

    if window.count(disk) == 4:
        score += 100
    elif window.count(disk) == 3 and window.count(EMPTY) == 1:
        score += 5
    elif window.count(disk) == 2 and window.count(EMPTY) == 2:
        score += 2

    if window.count(opp_disk) == 3 and window.count(EMPTY) == 1:
        score -= 4

    return score
```
